```python
# unique : unique values present in a column
# nunique : total number of unique values in a column
# value_counts : frequency of each unique value in a column

df['left'].unique()
```

```python
# Total attrition rate:

unique_labels= df['left'].unique()

count_left= df['left'].value_counts()
count_left
```
